```python
import pandas as pd
from sklearn.model_selection import train_test_split

data = pd.read_csv("processed/clean_data.csv")

data["year"] = data["title"].str.extract(r"\((\d{4})\)")
data["year"] = pd.to_numeric(data["year"], errors="coerce")
data["movie_age"] = 2026 - data["year"]
data["genres"] = data["genres"].str.split("|")
data["genre_count"] = data["genres"].apply(len)
```

```python
train_data, test_data = train_test_split(data, test_size=0.2, random_state=42)

movie_popularity = train_data.groupby("movie_id")["rating"].count().rename("movie_popularity")
movie_mean = train_data.groupby("movie_id")["rating"].mean().rename("average_rating")

train_data = train_data.merge(movie_popularity, on="movie_id").merge(movie_mean, on="movie_id")
test_data = test_data.merge(movie_popularity, on="movie_id", how="left").merge(movie_mean, on="movie_id", how="left")
test_data["movie_popularity"] = test_data["movie_popularity"].fillna(0)
test_data["average_rating"] = test_data["average_rating"].fillna(train_data["average_rating"].mean())
```

```python
train_data.to_csv("processed/featured_train.csv", index=False)
test_data.to_csv("processed/featured_test.csv", index=False)
print("✅ Fichiers créés.")
```
