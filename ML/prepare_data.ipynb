```python
import duckdb
import pandas as pd

con = duckdb.connect("../movielens_pipeline.duckdb")
data = con.execute("SELECT * FROM recommandation_prete").df()
con.close()

print(data.shape)
data.head()
```

```python
print("Doublons :", data.duplicated().sum())
data = data.drop_duplicates()
print("Valeurs manquantes :\n", data.isnull().sum())
```

```python
import os
os.makedirs("processed", exist_ok=True)
data.to_csv("processed/clean_data.csv", index=False)
print("✅ clean_data.csv créé.")
```
