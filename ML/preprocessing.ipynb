```python
import pandas as pd
from sklearn.preprocessing import MultiLabelBinarizer
import ast

train = pd.read_csv("processed/featured_train.csv")
test = pd.read_csv("processed/featured_test.csv")

train["genres"] = train["genres"].apply(ast.literal_eval)
test["genres"] = test["genres"].apply(ast.literal_eval)

mlb = MultiLabelBinarizer()
mlb.fit(train["genres"])

train_genres = pd.DataFrame(mlb.transform(train["genres"]), columns=mlb.classes_, index=train.index)
test_genres = pd.DataFrame(mlb.transform(test["genres"]), columns=mlb.classes_, index=test.index)

train = pd.concat([train, train_genres], axis=1).drop(columns=["title", "genres"])
test = pd.concat([test, test_genres], axis=1).drop(columns=["title", "genres"])
```

```python
X_train = train.drop(columns=["rating"])
y_train = train["rating"]
X_test = test.drop(columns=["rating"])
y_test = test["rating"]

X_train.to_csv("processed/X_train.csv", index=False)
X_test.to_csv("processed/X_test.csv", index=False)
y_train.to_csv("processed/y_train.csv", index=False)
y_test.to_csv("processed/y_test.csv", index=False)
print("✅ Prêt pour Jihad.")
```
